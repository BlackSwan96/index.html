```python
import numpy as np # Homework 2 Sample mean calculation
x= np.array([28.5,12.1,12.2,9.8,3.8,1.1,25.1,23.7,10.1,2.7,19.3,27.3,21.5,17.5,5.3,29.3,6.7,19.1,25.9,8.2,17.3])
xbar = np.mean(x);
print('xbar= {:.4f} for n = {:d} observations'.format(xbar,len(x)))
```

```python
import numpy as np # median
x = np.array([28.5,12.1,12.2,9.8,3.8,1.1,25.1,23.7,10.1,2.7,19.3,27.3,21.5,17.5,5.3,29.3,6.7,19.1,25.9,8.2,17.3])
n = len(x); print('number of data points=',n)
print('sorted array:',np.sort(x))
xmed = np.median(x);
print('xmed=',xmed)
```

```python
# regular mean calculation for Homework #2
x= (np.array([10,23,28,30,32,34,34,float("nan")]))
xbar = np.mean(x); print('regular mean xbar={:.4f}'.format(xbar))
xbar = np.nanmean(x); print('nanmean () xbar = {:.4}'.format(xbar))
```

```python
#trimmed mean
from scipy import stats
x = np.array([10,23,25,28,30,32,34,34,float("nan")])
xbar = stats.trim_mean(x,0.1);
print('trimmed mean xbar={:.4f}'.format(xbar))
```

```python
import numpy as np; import pandas as pd # VAriance and Standard Deviation
x = np.array([28.5,12.1,12.2,9.8,3.8,1.1,25.1,23.7,10.1,2.7,19.3,27.3,21.5,17.5,5.3,29.3,6.7,19.1,25.9,8.2,17.3])
xbar = np.mean(x)
d = x - xbar;
d2= d**2
stdframe = pd.DataFrame({'x':x, 'xbar':np.repeat(xbar,len(x)),'d':d,'d2':d2})
print(stdframe,'\n')
print("sumofdeviations=",np.sum(d),'\n')
s = np.std(x,ddof=1);
sig = np.std(x);
print('s, sig = ',s,sig)
```
